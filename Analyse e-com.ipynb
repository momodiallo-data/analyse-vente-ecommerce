```python
# @title
print("""                           ANALYSE DES VENTES E-COMMERCE
""")







print("""






CONTEXTE:

Cette étude analyse les ventes d'une entreprise e-commerce afin de mieux comprendre les
performances globales.



OBJECTIFS:

Identifier les produits les plus rentables, les meilleurs clients, les villes performantes et les
tendances de ventes.""")
```

```python
# @title
print("""

DONNÉES:


Les données contiennent des informations sur les clients, les produits, les prix, les quantités, les
villes et les dates de vente.



""")



import pandas as pd

data = {
    "client": ["Awa","Ibrahima","Khadija","Ousmane","Mariama","Abdou","Sokhna","Cheikh","Aminata","Babacar","Fatimata","Serigne","Awa","Ousmane","Mariama","Cheikh"],
    "ville": ["Dakar","Dakar","Thiès","Thiès","Dakar","Thiès","Dakar","Dakar","Thiès","Dakar","Dakar","Thiès","Dakar","Thiès","Dakar","Dakar"],
    "produit": ["Téléphone","Ordinateur","Casque","Téléphone","Casque","Ordinateur","Téléphone","Casque","Ordinateur","Téléphone","Casque","Téléphone","Ordinateur","Casque","Téléphone","Téléphone"],
    "prix": [200,800,50,200,50,800,200,50,800,200,50,200,800,50,200,200],
    "quantite": [3,1,5,2,3,2,4,6,1,5,4,3,1,2,2,3],
    "date": ["2024-01","2024-01","2024-02","2024-02","2024-03","2024-03","2024-01","2024-02","2024-03","2024-01","2024-02","2024-03","2024-04","2024-04","2024-04","2024-04"]
}

df = pd.DataFrame(data)
df
```

```python
# @title
print("1-NETTOYAGE DES DONNÉES")
print("""


Il s'agit là de vérifier que les données sont bien propres et prêtes à être analysées, en essayant de détecter
les données manquantes, les doublons ,les valeurs bizarres ,...


""")
```

```python
# @title
print("""Infos sur les données:

""")
df.info()
```

```python
# @title
print("""Doublons:

 """)
df.duplicated().sum()
```

```python
# @title
print("""Descriptions:

""")
df.describe()
```

```python
# @title
print("On peut conclure que les données sont bien propres et prêtes à l'analyse")
```

```python
# @title
print("2-AJOUT DES COLONNES REVENU ET MOIS ")
```

```python
# @title
df["revenu"] = df["prix"] * df["quantite"]
df
```

```python
# @title
df["date"] = pd.to_datetime(df["date"])
df["mois"] = df["date"].dt.month
df
```

```python
# @title
print("3-ANALYSE CLIENTS")
print("""

Nous analysons les clients afin d’identifier ceux qui génèrent le plus de revenus. Cela permet de
cibler les profils les plus rentables.
""")
```

```python
# @title
print("-Top client par revenu")
```

```python
# @title
df.groupby("client")["revenu"].sum().sort_values(ascending=False).head(3)

```

```python
# @title
print("-Fréquence d'achat par client")
```

```python
# @title
df.groupby("client")["revenu"].count().sort_values(ascending=False)

```

```python
# @title
print("""
Abdou et Awa se démarquent clairement par leur contribution au chiffre d'affaires



""")
print("4-ANALYSE PRODUIT ")
```

```python
# @title
print("""Nous étudions les produits afin de déterminer lesquels sont les plus vendus et les plus rentables.

""")
print("-Produit le plus rentable")
```

```python
# @title
df.groupby("produit")["revenu"].sum().sort_values(ascending=False)
```

```python
# @title
print("produit le plus rentable:",df.groupby("produit")["revenu"].sum().idxmax())
```

```python
# @title
print("-Produit le plus vendu")
```

```python
# @title
df.groupby("produit")["quantite"].sum().sort_values(ascending=False)
```

```python
# @title

print("produit le plus vendu:",df.groupby("produit")["quantite"].sum().idxmax())
```

```python
# @title
print("-Évolution des ventes par produit")
```

```python
# @title

df.groupby(["produit","mois"])["revenu"].sum().unstack()
```

```python
# @title
print("visualisation des ventes par produit et par mois")
```

```python
# @title
df.groupby(["produit","mois"])["revenu"].sum().unstack().plot()
```

```python
# @title

print("Interprétation:")
```

```python
# @title



print("""Le graphique de l'évolution des ventes par produit et par mois montre que le produit téléphone
a été le plus rentable au mois de janvier,que les ventes au mois de février sont pratiquement
négligeables par rapport aux autres étant très faibles, mais le casque est celui qui se vend le mieux
quand même.
Au mois de mars, l'ordinateur a été le produit qui génére le plus de revenus,tandis qu'au mois d'avril
les ventes sont plus faibles par rapport aux mois de janvier et mars,avec le téléphone qui se vend légèrement
 mieux que l'ordinateur.""")
```

```python
# @title
print("6-ANALYSE GEOGRAPHIQUE")
print("""

Nous analysons les ventes par ville afin d’identifier les zones les plus performantes.


-Produits vendus par ville en terme de revenus
""")
```

```python
# @title
df.groupby(["ville","produit"])["revenu"].sum().unstack()
```

```python
# @title
print("-Revenu par ville")
```

```python
# @title
df.groupby("ville")["revenu"].sum()

```

```python
# @title
print("-Produits vendus par ville")
df.groupby(["ville","produit"])["quantite"].sum()
```

```python
# @title
print("""Dakar génére plus de revenus que Thiès et les téléphones marchent le mieux  là bas.
Les ordinateurs générent plus de revenus à Thiès.




""")
print("7-VISUALISATION")

import matplotlib.pyplot as plt
df.groupby("mois")["revenu"].sum().plot(kind="bar")
plt.title("revenu par mois")
plt.xlabel("mois")
plt.ylabel("revenu")
plt.show()
```

```python
# @title

import matplotlib.pyplot as plt
df.groupby("produit")["revenu"].sum().plot(kind="bar")
plt.title("revenu par produit")
plt.xlabel("produit")
plt.ylabel("revenu")
plt.show()
```

```python
# @title
import matplotlib.pyplot as plt
df.groupby(["ville","produit"])["quantite"].sum().plot(kind="bar")
plt.title("produit par ville")
plt.xlabel("ville")
plt.ylabel("produit")
plt.show()
```

```python
# @title
print("""INTERPRÉTATIONS:
-Le graphique de revenu par mois montre que les ventes évoluent à dents de scie au cours
du temps.
Les revenus atteignent leur pic au mois de janvier, puis rechutent considérablement au mois
de février.
Les ventes s'élèvent une nouvelle fois au mois de mars avant de retomber en avril.

-L'étude des ventes des différents produits nous permettent de constater que les téléphones
sont ceux qui générent le plus de revenus, et devancent de peu ceux des ordinateurs malgré
la faiblesse des ventes en terme de quantité.

 -L'étude du 3éme graphique montre les téléphones sont les produits qui se vendent le plus à
  Dakar en terme de quantité, tandis que les casquent sont les plus vendus à Thiès.
""")
```
